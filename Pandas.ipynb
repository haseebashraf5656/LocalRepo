```python
import pandas as pd
import numpy as np
data ={
    'Name':['Haseeb', 'Ali', 'Ahmed', 'Amna', 'Haseeb'],
    'Marks':[20, 30, pd.NA, 10, 20],
    'gender':['M', 'M', 'M', 'F', 'M']
}
df = pd.DataFrame(data)
print(df)
print(df.duplicated())
average = df['Marks'].mean()
print(average)
df['Marks'] = df['Marks'].fillna(average)
print(df)
df.drop_duplicates()
df.drop(['Marks'], axis = 1)
df.to_csv('data.csv')
```

```python
import pandas as pd
import numpy as np
data ={
    'Name':['Haseeb', 'Ali', 'Ahmed', 'Amna', 'Haseeb', 'Naveed'],
    'Marks':[20, 30, pd.NA, 10, 1000, 200],
    'gender':['M', 'M', 'M', 'F', 'M', 'M']
}
df = pd.DataFrame(data)
Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Marks'] < lower_bound) | (df['Marks'] > upper_bound)]
print(outliers)
df_clean = df[(df['Marks'] >= lower_bound) & (df['Marks'] <= upper_bound)]
print(df_clean)
```

Sales > 8000

```python
import pandas as pd

data = {
    "EmployeeID": [101, 102, 103, 104, 105],
    "EmployeeName": ["Haseeb", "Ali", "Ahmed", "Sara", "Usman"],
    "Department": ["Sales", "Marketing", "Marketing", "Sales", "Finance"],
    "Sales": [50000, 30000, 0, 45000, 60000],
    "Salary": [70000, 90000, 55000, 68000, 75000],
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Faisalabad"]
}

df = pd.DataFrame(data)
print(df)
# print(df.info())
# print(df.describe())


print("Employees from marketing department")
new_data = df[(df['Department'] == 'Marketing') & (df['Sales'] > 8000)]
print(new_data)

print("Salary Sorted Data")
df['Salary'] = df['Salary'].sort_values()
print(df)

print("Adding New column comission")
df['Comission'] = df['Sales']*0.05
print(df)

print("Adding new row")
df.loc[len(df)] = [110, 'Ahmed', 'Sales', 1000, 40000, 'Lahore', 900]
print(df)
```
